**1) JGR Driver Stats 2026 Visualization**

```python
# Import libraries
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt

# Import NASCAR data / CSVs
loopdata_2026 = pd.read_csv("cup_loopdata_2026_r1-r23.csv")
cup_rr_results_app = pd.read_csv("cup_results_updated_8_10_2026.csv", encoding="latin1")
# Note: cup_rr_results_app contains mixed data types Columns (1,2,6,8).
# mixed data types columns (1,2,6,8) are converted to integers during data cleaning below.
```

**2) Inspect DF**

```python
# Clean & Filter Data
# Clean cup_results_app['Driver'] whitespace
cup_rr_results_app['Driver'] = cup_rr_results_app['Driver'].str.strip()


# Filter to 2026
cup_rr_results_2026 = cup_rr_results_app[cup_rr_results_app['season']==2026].copy()

# Fixed Mixed datatypes
# DtypeWarning: Columns (1,2,6,8) have mixed types.
cols_to_int = ["St", "#", "Laps", "Led"]
cup_rr_results_2026[cols_to_int] = cup_rr_results_2026[cols_to_int].astype(int)

# Convert ['date] to datetime format
cup_rr_results_2026['date'] = pd.to_datetime(cup_rr_results_2026['date'])

# Divide the season into monthly phases: using "cut()"
cup_rr_results_2026["phase"] = pd.cut(
    cup_rr_results_2026["date"],
    bins=[
        pd.Timestamp("2026-02-14"),  # boundary 1
        pd.Timestamp("2026-03-15"),  # boundary 2
        pd.Timestamp("2026-04-15"),  # boundary 3
        pd.Timestamp("2026-05-15"),  # boundary 4
        pd.Timestamp("2026-06-15"),  # boundary 5
        pd.Timestamp("2026-07-15"),  # boundary 6
        pd.Timestamp("2026-08-15"),  # boundary 7
    ],
    labels=[
        "Feb15 - Mar15",
        "Mar16 - Apr15",
        "Apr16 - May15",
        "May16 - Jun15",
        "Jun16 - Jul15",
        "Jul16 - Aug15"
    ]
)


# Create Lists of Driver/Teams To use as filters later
# Driver/Teams List
jgr_drivers = [
    'Denny Hamlin',
    'Ty Gibbs',
    'Christopher Bell',
    'Chase Briscoe']
hmr_drivers = [
    'Kyle Larson',
    'Alex Bowman',
    'William Byron',
    'Chase Elliott']
twenty311_drivers=[
    "Bubba Wallace",
    "Tyler Reddick",
    "Riley Herbst"]
spire_drivers=[
    "Carson Hocevar",
    "Daniel Suarez", # Check Spelling/enconding in CSV
    "Michael McDowell"]
trackhouse_drivers =[
    "Shane Van Gisbergen",
    "Ross Chastain",
    "Connor Zilisch"
]

# Create DFs for each respective team
jgr_df = cup_rr_results_2026[cup_rr_results_2026["Driver"].isin(jgr_drivers)].copy()
hmr_df = cup_rr_results_2026[cup_rr_results_2026["Driver"].isin(hmr_drivers)].copy()
t3xi_df = cup_rr_results_2026[cup_rr_results_2026["Driver"].isin(twenty311_drivers)].copy()
trackhouse_df = cup_rr_results_2026[cup_rr_results_2026["Driver"].isin(trackhouse_drivers)].copy()
```

**3) Set the DF in order by race_number**

```python
# Set the DF in order by race_number
jgr_df = jgr_df.sort_values("race_number").reset_index(drop=True)

# Create a list of the tracks in order by race number
tracks_in_order = jgr_df[["site", "race_number"]].drop_duplicates().sort_values("race_number")["site"].to_list()

# Visually Inspect race order and site are correct
for i, track in enumerate(tracks_in_order):
    print(i+1, track)
```

**4) Create Line Chart**

```python
# Create Line Chart
fig = px.line(
    jgr_df,  # Data / DataFrame and Columns to use
    x="race_number", # x-axis
    markers=True,
    y="Pts",        # y-axis
    color="Driver", # Grouping
    color_discrete_map={
    "Denny Hamlin": "#0077B3",       # Progressive Blue
    "Ty Gibbs": "#7CB701",            # Monster Energy Green
    "Chase Briscoe": "#ED3A42",       # Bass Pro Shops Red
    "Christopher Bell": "#FEBD17"},     # DeWalt Yellow
    template="plotly_dark") # theme


# Update grid-lines
fig.update_xaxes(
    gridcolor="rgba(255,255,255,0.02)",
    tickvals=list(range(1, len(tracks_in_order)+1)),  # Use len(tracks_in_order) for x-axis
    ticktext=tracks_in_order) # Text on x-axis to display track names


fig.update_yaxes(
    gridcolor="rgba(255,255,255,0.02)")

fig.update_layout(
    width=1200,
    height=500)

fig.show()
```
